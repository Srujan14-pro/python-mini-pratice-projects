```python
num = int(input("Enter your num: "))

print("========== NUMBER REPORT ==========")

total_sum = 0

count_even = 0
count_odd = 0
count_prime = 0
count_not_prime = 0
sumof_even = 0
sumof_odd = 0
sumof_prime = 0

for i in range(1, num + 1):

    # Even / Odd
    if i % 2 == 0:
        sumof_even += i
        count_even += 1
    else:
        sumof_odd += i
        count_odd += 1

    if num==0 or num==1:
        count_not_prime += 1
    elif num>1:
        for i in range(2,num):
            if num%i==0:
                count_not_prime += 1
    else:
        sumof_prime += i
        count_prime += 1






    # Total
    total_sum += i


print(f"The total nums are {num}")
print(f"The even nums are {count_even}")
print(f"The odd nums are {count_odd}")
print(f"The not prime nums are {count_not_prime}")
print(f"The prime nums are {count_prime}")
print(f"The sum of nums is {total_sum}")
print(f"The sum of even nums is {sumof_even}")
print(f"The sum of odd nums is {sumof_odd}")
print(f"The sum of prime nums is {sumof_prime}")
```
